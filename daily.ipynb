        - why decoupling storage and compute or seperate processing layers (like Snowflake 3-layer architecture) for cloud DWH changes the game: 
            - storage is persistent and cheap
            - elasticity in cloud computing: 
                - easy explanation: when traffic increases, the system provisions additional resources. When demand drops, it scales back down -> automatic scaling
                - adapts infrastructure in real time to match actual workload demands - unlike traditional infra that requires manual update
                - operates in 2 scaling direcions:
                    - vertical (scale up/down): adding more power to existing resources, like increasing CPU or memory on a virtual machine.
                    - horizontal (scale out/in): adding or removing entire instances, like spinning up additional servers to handle traffic.
            - isolate workloads and scale each layer independently

        - distributed computing systems: 
            - explanation: the method of making multiple computers work together to solve a common problem
            - node:  a single point or building block (either hardware or software such as a computer, server, or virtual machine) in a network

        - how DuckDB levels up: 
            - single-node (the computation is happening on one machine) -> simpler, cheaper, less infrastructure; can still use multiple CPU cores on that machine
            - federated: data/computation can exist across different locations (e.g., local machine + cloud) while being accessed through one DuckDB-based workflow
            - while traditional db like PostgreSQL needs a seperate db server to process your sql query,  DuckDB runs the query inside your application/process
            --> allowing DuckDB's query engine to work with data across local and cloud environments
        
        - ClickHouse: distributed, column-oriented OLAP db
            - designed for very fast analytical queries, especially large volumes of event data
            - particularly strong for streaming / near real-time workloads
            - can ingest streams of events and aggregate them on the fly
            - unlike DuckDB, ClickHouse is designed for distributed, high-performance analytics, while DuckDB emphasizes local/embedded simplicity

        - dbt in the T (Transformation):
            - dbt is a transformation tool that turns raw data into analytics-ready models; works on top of your warehouse/database
            - essentially SQL + Jinja + version control + testing/documentation
            - '{{ ref() }}' → references another model and creates dependencies
        
        - orchestration:
            - basically means make things automatic: coordinate and automate the data pipeline tasks such as scheduling, order/dependencies, retries, monitoring, etc
            - Airflow: use DAGs (Directed Acyclic Graphs) to define workflows
                - DAGs: a mathematical structure made of nodes (points) connected by directed edges (one-way arrows like a one-way street) that never form a loop

        - version control + CI/CD:
            - why need Git?
                - tracks changes + allows rollbacks + collaboration
            - linting: checks code for syntax/style issues (ruff, black, sqlfluff)
            - CI: automatically test/validate changes
            - CD: automatically deploy approved changes
